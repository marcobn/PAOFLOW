```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
conductor_bulk = np.loadtxt("./output/paoflow/conductance_bulk.dat", skiprows=1)
conductor_defect = np.loadtxt("./output/paoflow/conductance_defect.dat", skiprows=1)
energies_bulk = conductor_bulk[:, 0]
energies_defect = conductor_defect[:, 0]
transmissions_bulk = conductor_bulk[:, 1]
transmissions_defect = conductor_defect[:, 1]
plt.plot(energies_defect, transmissions_defect, label="Defect")
plt.plot(energies_bulk, transmissions_bulk, label="Bulk")
plt.xlabel("Energy (eV)")
plt.ylabel("Transmission")
plt.title("Transmission vs Energy for Al Defect Conductor")
plt.grid()
plt.legend()
plt.show()
```

```python
conductor_bulk = np.loadtxt("./output/paoflow/doscond_bulk.dat", skiprows=1)
conductor_defect = np.loadtxt("./output/paoflow/doscond_defect.dat", skiprows=1)
energies_bulk = conductor_bulk[:, 0]
energies_defect = conductor_defect[:, 0]
transmissions_bulk = conductor_bulk[:, 1]
transmissions_defect = conductor_defect[:, 1]
plt.plot(energies_defect, transmissions_defect, label="Defect")
plt.plot(energies_bulk, transmissions_bulk, label="Bulk")
plt.xlim(-1.2, 1.2)
plt.xlabel("Energy (eV)")
plt.ylabel("Transmission")
plt.title("Transmission vs Energy for Al Defect Conductor")
plt.grid()
plt.legend()
plt.show()
```
