```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
bands = np.loadtxt('dense/bands_0.dat')
dos = np.loadtxt('dense/dosdk_0.dat')
pdos = np.loadtxt('dense/pdosdk_sum_0.dat')
kappa = np.loadtxt('dense/kappagauss_0.dat')
sigma = np.loadtxt('dense/sigmagauss_0.dat')
seebeck = np.loadtxt('dense/Seebeckgauss_0.dat')
pf = np.loadtxt('dense/PFgauss_0.dat')
```

```python
k = bands[:, 0]
bands = bands[:, 1:]

plt.figure(figsize=(7, 5))

for ib in range(bands.shape[1]):
    plt.plot(k, bands[:, ib], color="black", linewidth=0.7)

plt.xlabel("k-point index")
plt.ylabel("Energy / eigenvalue")
plt.title("Band structure")
plt.grid(alpha=0.25)

plt.tight_layout()
plt.show()
```
